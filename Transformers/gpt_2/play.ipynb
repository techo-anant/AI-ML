````python
with open('input.txt', 'r') as f:
    text = f.read()
data = text[:1000] # first 1000 characters
print(data[:100])
````

````python
import tiktoken
enc = tiktoken.get_encoding('gpt2')
tokens = enc.encode(data)
print(tokens[:24])
````

````python
# process these token sequences and feed them to a transformer
import torch
buf = torch.tensor(tokens[:24 + 1])
x = buf[:-1].view(4, 6) # input (B, T)
y = buf[1:].view(4, 6) # targets i.e. labels
print(x)
print(y)
````
